# 1. HTTP基本原理
- URI: Uniform Resource Identifier，统一资源标识符
- URL: Uniform Resource Locator，统一资源定位符

URL是URI的子集，如 `https://github.com/favicon.ico`，它既是一个URI也是一个URL
URL格式：
schema://[username:password@]hostname[:port][/path][;paremeters][?query][#fragment]
- `schema`: 也可以叫 `protocol`，代表协议，比如 `http`, `https`, `ftp`等
- `username`, `password`: 用户名和密码。某些URL需要提供用户名和密码才能访问
- `hostname`: 主机地址。也可以是域名或ip地址。比如`www.baidu.com`
- `port`: 端口。大部分端口其实是和协议绑定的。比如`https`: 443, `http`: 80等，所以需要指定端口的情况比较少。
- `path`: 路径。表示网络资源在服务器中的指定地址。
- `parameters`: 参数。用于来指定访问某个资源时的附加信息。但是现在`parameters`用的很少，所以很多人会把后面的`query`部分称为参数，甚至把二者混用。
- `query`: 查询。用来查询某类资源，如果有多个查询，用`&`分割。
- `fragment`: 片段。用来对资源描述的补充，可以理解为资源内部的书签。目前有两个应用：现代前端框架如`Vue`，`React`用他做路由管理；HTML页面中，用作锚点，控制一个页面打开时自动下滑滚动到某个特定位置。

# 2. HTTP和HTTPS
两者的区别就在于SSL层。对于HTTPS而言，所有的传输内容都经过了SSL的加密。
SSL的作用有2点：
1. 建立一个信息安全通道，保证数据传输的安全性
2. 确认网站的真实性，凡是使用了HTTPS的网站，都可以通过单机浏览器地址栏的锁头标志来查看网站认证之后的真实值信息。此外还能通过CA机构颁发的安全证书来查询。

# 3. Request
分为4部分
1. Method
2. URL
3. Headers
4. Body

## 3.1. Method
用于标识请求客户端请求服务器的方式，常见的就是`GET`和`POST`，其区别在于
1. `GET`中的参数包含在URL里，数据可以从URL中看到；而`POST`通过表单传输，数据包含在Body中
2. `GET`最多只能有1024B的数据，而`POST`无限制

| method| 描述|
|----|----|
| GET | 请求页面，并返回页面内容|
| HEAD | 类似于GET，但是返回的响应中没有具体内容。用于获取报头 |
| POST | 大多用于提交表单或上传文件，数据包含在Body中 |
| PUT | 用客户端传向服务器的数据取代制定文档中的内容 |
| DELETE | 请求服务器删除指定的页面 |
| CONNECT | 把服务器当做跳板，让服务器代替客户端访问其他页面 |
| OPTIONS | 允许客户端查看服务器性能 |
| TRACE | 回显服务器收到的请求。主要用于测试或诊断 |

## 3.2. URL
用于唯一确定客户端想请求的资源。前面已经介绍过了，此处略过

## 3.3. Headers
用来说明服务器要使用的附加信息
1. Accept: 用于指定客户端可接受哪些类型的信息
2. Accept-Language: 用于指定客户端可接受的语言类型
3. Accept-Encoding: 用于指定客户选可接受的编码
4. Host: 用于指定请求资源的主机IP和Port，其内容为请求URL的原始服务器或网关的位置。从HTTP1.1开始，必须包含此内容
5. Cookie: 也常用复数形式Cookies，这是网站为了辨认用户，进行会话跟踪而存储在用户本地的数据。其主要功能是维持当前会话。常见的应用就是维持登录状态，下次访问时自动登录
6. Referer: 用于标识Request是从哪个页面发过来的
7. User-Agent: 这是一个特殊的字符串头，可以让服务器识别客户端使用的操作系统及版本，浏览器及版本信息等。爬虫中可以伪造此信息
8. Content-Type: 也叫MIME类型，互联网媒体类型，用来表示Request的媒体类型信息。比如text/html代表HTML格式，application/json表示JSON类型

## 3.4. Body
一般用于承载POST中的表单数据。


# 4. Response
即由服务器返回给客户端的内容。分为3部分
1. Status code
2. Headers
3. Body

## 4.1. Status Code
表示服务器的响应状态。如200表示正常响应，404代表页面没找到

## 4.2. Headers
包含了服务器对Request的应答信息。常见的如下
1. Date: 标识Response产生的时间
2. Last-Modified: 用于指定资源最后的修改时间
3. Content-Encoding: 用于指定响应内容的编码
4. Server: 包含服务器的信息，例如名称，版本号等
5. Content-Type: 文档类型。
6. Set-Cookie: 告诉浏览器需要将此内容放在Cooke中，下次请求时携带上
7. Expires: 指定Response的过期时间，可以让代理服务器或浏览器将加载的内容更新到缓存中，当再次访问相同内容时，可以直接从缓存中加载

## 4.3. Body
Response的正文数据都在Body中，需要解析的内容就是Body

# 5. 爬虫的基本原理
简单而言，爬虫就是获取网页并提取和保存信息的自动化程序

1. 获取网页：即拿到网页的源代码
2. 提取信息：从拿到的源代码中提取出我们需要的信息
3. 保存数据
4. 自动化程序：即将上述的操作转化为程序

有一个常见的问题，现在越来越多网页使用AJAX，前端模块化工具构建，HTML只是一个空格，由JS来渲染出来。这就需要一些库来模拟JS渲染的过程，比如 `Selenium`，`Splash`，`Pyppeteer`，`Playwright`等

# 6. Session和Cookie
首先我们需要知道，HTTP是一个无状态协议，即对事务处理是没有记忆能力。由此衍生出了保持连接状态的技术，即Session和Cookie。
Session在服务器端，用来保存用户的信息。
Cookie在客户端，或者说浏览器端。
浏览器会在访问时自动带上Cookie，服务器通过识别Cookie来鉴定是哪个用户。
所以在爬虫中我们么通常会直接使用Cookie而不是用用户名和密码来登录

## 6.1. Cookie
对于Cookie，大部分情况下会有复数个，具有下列属性
- Name: 名称。一旦被创建，则不可更改
- Value: 值。如果是Unicode字符，则需要为字符编码。如果是二进制数据，则使用Base64编码
- Domain: 指定可以访问该Cookie的域名。如设置为.zhihu.com的话，则表示所有以zhihu.com结尾的域名都可以访问该cookie
- Path: 使用路径
- Max-Age: 失效时间，单位为s。如果是负数，则表示在关闭浏览器时失效。通常和Expires一起使用。
- Size: Cookie的大小
- HTTP: httponly属性。如果为true，则只有HTTP headers中才会有此信息
- Secure: 是否允许仅使用安全协议传输Cookie，默认是False

# 7. 代理
在做爬虫时，我们可能会遇到一开始程序顺利运行，但是一段时间后，就会发生403Forbidden错误，此时打开网页就会看到“您的IP访问频率太高”这样的提示。
这是因为一些网站采取了防爬措施，比如服务器检测到某个IP在单位时间内的请求次数到达阈值的话，就会直接拒绝提供服务。
对此的结局方案就是用某种方法把我们的IP伪装一下，这就是代理服务器（Proxy Server）。
